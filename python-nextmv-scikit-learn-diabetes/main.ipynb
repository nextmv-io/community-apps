# `nextmv-scikit-learn` interfacing examples

In this notebook, you will learn how to use the `nextmv_sklearn` library to
interface with various `sklearn` models.

## Data

Let’s load some data to work with the different models.

```python
from sklearn.datasets import load_diabetes

X, y = load_diabetes(return_X_y=True)
```

## Dummy Regressor

Add the necessary imports.

```python
from nextmv_sklearn.dummy import (
    DummyRegressor,
    DummyRegressorOptions,
    DummyRegressorSolution,
    DummyRegressorStatistics,
)
```

Initialize the options, which are of type `nextmv.Options`.

```python
dum_opt = DummyRegressorOptions().to_nextmv()
```

Create and fit the model (type `sklearn.dummy.DummyRegressor`) with the options. Make
a basic prediction.

```python
dum_reg = DummyRegressor(dum_opt)
dum_fit = dum_reg.fit(X, y)
dum_fit.predict(X[:1])
```

Get the base statistics of the model. We encourage that you customize the statistics.

```python
dummy_stats = DummyRegressorStatistics(dum_fit, X, y)
dummy_stats.to_dict()
```

The fitted model is considered a "solution". Create a solution object from the
model and export it to a file.

```python
import nextmv

dum_sol = DummyRegressorSolution.from_model(dum_fit)
output = nextmv.Output(
    options=dum_opt,
    solution=dum_sol.to_dict(),
    statistics=dummy_stats,
)
dum_file_path = "dummy_regressor.json"
nextmv.write_local(output, dum_file_path)
```

At any moment, the solution (fitted model) can be loaded from a file that holds the correct
information. Load the solution, create a fitted model from it, and make a
basic prediction.

```python
nm_input = nextmv.load_local(path=dum_file_path)
dum_sol_2 = nextmv.from_dict(nm_input.data["solution"])
dum_fit_2 = dum_sol_2.to_model()
dum_fit_2.predict(X[:1])
```
